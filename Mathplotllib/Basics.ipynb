**Basics of Mathplotlib**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(1,10,11)
x
y=x**2
y
```

```python
plt.plot(x,y)
plt.title("Normal")
plt.xlabel("Age")
plt.ylabel("Height")
```

**subplot**

```python
plt.subplot(2,2,1)
```

```python
plt.subplot(2,2,1)
plt.plot(y,x**5)
plt.subplot(2,2,2)
plt.plot(x,y) #just write ehat you want to plot under the subplot section 
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)#2,2 signifies the number of column and rows and the 4 boxes are created and we can plot them 
plt.plot(y,x**2)
```

**Object Oriented way**

```python
fig=plt.figure(figsize=(12,8))
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2=fig.add_axes([0.4,0.2,0.4,0.6])
axis2.plot(y,x)
```

**Types of plot**

```python
plt.scatter(y,x)
```

```python
a=[10,20,300,100,2,3,67]
plt.hist(a)
```

```python
plt.boxplot(x)
```

**SAVING A PLOT**


```python
fid=plt.figure(figsize=(12,8))
axis=fid.add_axes([0.1,0.1,0.8,0.8])
axis.plot(x,y)
plt.savefig("demo.png")
```
