## ARRAY OPERATIONS
1. Arithmetic operations
2. Broadcasting
3. Deep and Shallow copy
4. Matrix operations


```python
import numpy as np
# Array operations
ar1=np.arange(1,6)
ar2=np.arange(6,11)
print(ar1)
print(ar2)
print("Addition: ",ar1+ar2)
print("Substraction: ",ar1-ar2)
print("Divition: ",ar1/ar2)
print("Floor Division: ",ar1//ar2)
print("Multiplication: ",ar1*ar2)
print("Power:  ",ar1**ar2)
print("All the arithmettic operations are completed and there should bew equal no of elemnts in both the elemnts or else there will be broadcasting error we will see more in pandas")
```

```python
#broadcasting
#broadcasting is use when u wanbt to add single elemt or something to rest of the multiple data in the array
import numpy as np
arr= np.array([10,20,30,40]) 
print(arr*41)
print(arr+10)
print(arr//10)
```

```python
#deep and shallow copy
arr=np.arange(1,21)
arr
```

```python
a=arr[0:9]
a=a*10
print(a)
```

```python
b=a
b[0]=100 #deepcopy
a
```

```python
#shallow copy
c=a.copy()

c[3]=100
print(c)
print(a)
```

```python
#Matrixoperations
mat1=np.array([[1,2],[3,4]])
mat2=np.array([[5,6],[7,8]])
mat1
```

```python
mat2
```

```python
print(mat1 @ mat2 )#matrix multiplication
print(mat1-mat2)
```

## Advance array Manipulation
1. Stacking array
2. splitting array 

```python
a1=np.array([1,2,3,4])
b1=np.array([5,6,7,8])
a1
b1
```

```python
np.vstack((a1,b1))
```

```python
np.hstack((a1,b1))
```

```python
np.column_stack((a1,b1))
```

```python
j=np.arange(1,31).reshape(5,6)
print(j)
#plitting
```

```python
k=np.vsplit(j,5)
k
```

```python
l=np.hsplit(j,3)
print(l)
```
