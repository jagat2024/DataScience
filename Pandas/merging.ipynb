## MERGING AND CONCATENATION

```Merging all part```

````python
import numpy as np 
import pandas as pd

#1st dataframe
Employees=pd.DataFrame({
    'Emp_id':[1,2,3,4,5],
    'Emp_name':['Jagat','Sagnik','Ankit','Joydeep','Harshit'],
    'Department':['Senior Developer','Ml-senior Director','Supervisior','Backend Senior Developer','Supervisior']
})
#2nd dataframe
salary=pd.DataFrame({
    'Emp_id':[1,2,3,6,7],
    'Salary':[100000,120000,96000,100000,92000],
    'Bonus':[5000,3200,4500,2000,3400]
})
Employees
````

````python
salary
````

````python
#merging 
pd.merge(Employees,salary)#ondefault it take how ='inner'
````

````python
pd.merge(Employees,salary,how='inner')
````

````python
pd.merge(Employees,salary,how='outer')#it joins on basis of columns and make the places nan where the fields  are empty
````

````python
pd.merge(Employees,salary,how='left')#joins on basis of employees 
````

````python
pd.merge(Employees,salary,how='right')#on the basis of salary
````

```concatenation```

````python
#dataframe
df1=pd.DataFrame({
    'A':['A0','A1','A2'],
    'B':['B0','B1','B2'],
    'C':['C0','C1','C3']
})
#dataframe 2
df2=pd.DataFrame({
    'A':['A3','A4','A5'],
    'B':['B3','B4','B5'],
    'C':['C3','C4','C5']
})
df1
````

````python
df2
````

````python
pd.concat([df1,df2])#dafault concat columnwise
````

````python
pd.concat([df1,df2],axis=1)#row concat columnwise axis=1
````

```Joining```

````python
df3=pd.DataFrame({'Name':["Alice","Bob","joye"]},index=[1,2,3])
df4=pd.DataFrame({'Salary':[10000,9000,8000]},index=[2,3,4])
df3
````

````python
df4
````

````python
df3.join(df4)#default it takes inner
````

````python
df3.join(df4,how='outer')#now outer will take it all
````
