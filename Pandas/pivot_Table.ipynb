## Pivot Tables


```python
import numpy as np
import pandas as pd
```

```python

data={'Date':pd.date_range('2023-04-01',periods=20),
'Product':['A','B','C','D','E']*4,
'Region':[ 'East','west','North','South','East','west','North','South','East','west','North','South','East','west','North','South','East','west','North','South'],
'Sales':np.random.randint(10,1000,20),
'Units':np.random.randint(10,100,20),
'Rep':['John','Mary','Bob','Alice','John','Mary','Bob','Alice','John','Mary','Bob','Alice','John','Mary','Bob','Alice','John','Mary','Bob','Alice']
}
df=pd.DataFrame(data)
df
```

```python
df['month']=df['Date'].dt.month_name()
df
```

```python
df['Quater']='Q'+df["Date"].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values='Sales',index="Region",columns='Product')
```

```python
pd.pivot_table(df,values="Units",index="Region",columns='Product',aggfunc='median')
```

```python
pd.pivot_table(df,values=["Sales","Units"],index="Region",columns='Product',aggfunc='median')
```

```python
pd.crosstab(df['Units'],df['Region'])
```

```python
pd.crosstab(df['Units'],df['Sales'])
```
