## Groupby and Aggregation


````python
import numpy as np
import pandas as pd

#dataframe 1
data={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['s1','s1','s2','s2','s1','s2','s2','s1'],
    'Sales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01',periods=8)
}
cat=pd.DataFrame(data)
cat
````

````python

f=cat.groupby('Category')['Sales']
for i,v in f:
    print(i)
    print(v)

````

````python
#Group by category and say sum of sales
f=cat.groupby('Category')['Sales'].sum()
f
````

````python
#Group by multiple columns and say the sales
k=cat.groupby(['Category','Store'])['Sales'].sum()
k
````

## Aggregation

``` aggregate function use dto combine and provide all the statistical value in one time except the mode operation in pandas lets see the examples in the next step```

````python
print(cat['Sales'].mean())
````

````python
print(cat['Sales'].mode())
````

````python
print(cat['Sales'].max())
````

````python
print(cat['Sales'].min())
````

````python
print(cat['Sales'].var())
````

````python
print(cat['Sales'].std())
````

````python
cat['Sales'].agg(['max','min','mean','median','var','std'])
````
