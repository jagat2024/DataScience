````python
#CREATING A DATAFRAME USING DICTIONARY
DATA=  {"NAME": ['John','Anna','Peter','Linda'],
        "AGE": [26,34,23,24],
        "SALARY": [30000,40500,23000,12000],
        "CITY": ["Newyork","Paris","Berlin","London"]}
k=pd.DataFrame(DATA)
k

#CREATING A DATAFRAME USING LIST BUT DRAW BACK IS THAT WE NEED TO PUT THE COLUMNS MANUALLY
LIST = [
    ['John', 26, 30000, 'Newyork'],
    ['Anna', 34, 40500, 'Paris'],
    ['Peter', 23, 23000, 'Berlin'],
    ['Linda', 24, 12000, 'London']
]

DF = pd.DataFrame(LIST, columns=["NAME","AGE","SALARY","CITY"])
DF
````

````python
#Selection and indexing in datframe of the colums 
print(f"{DF["NAME"]}") #one column
print("===================================")
#for multiple columns 
print(f"{DF[["NAME","AGE"]]}")
#adding of colums 
DF["Designation"]=["HR","ENGINEER","DOCTOR","ENGLISH"]
# DF["DESIGNATION"]=["HR","ENGINEER","DOCTOR","ENGLISH"]
print(DF)
# deleting a column
DF.drop('Designation',axis=1)
````

````python
#for selecting a row 
print(DF)
print(" \n ")
DF.loc[[0,1]]

````
