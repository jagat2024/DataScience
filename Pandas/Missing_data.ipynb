```MISSING DATA MANIPLULATION```


````python
import numpy as np
import pandas as pd
````

````python
DATA={'A':[1,2,np.nan,4,5],
    'B':[np.nan,2,3,4,5],
    'C':[1,2,3,np.nan,np.nan],
    'D':[1,np.nan,np.nan,np.nan,5]
}
df=pd.DataFrame(DATA)
````

### missing data evaluation


````python
df
````

````python
df.isna()
````

````python
df.isna().sum()
# count the number of missing value in each column and give the value
````

````python
df.isna().any()
````

### Removing missing value


````python
df.fillna(df.mean())
````

````python
values={'A':0,'B':100,"C":1000,"D":10000}
df.fillna(values)#agar yaha pe inplace true likh dunga toh main data mai changes aa jayenge 
````
